```python
orders['order_date']=pd.to_datetime(orders['order_date'])
```

```python
lastorders=orders.groupby("customer_id")['order_date'].max()
lastorders=lastorders.reset_index()
lastorders
```

```python
lastorders["Recency"]=(orders['order_date'].max()-lastorders['order_date']).dt.days
lastorders
```

```python
Frequency=orders.groupby("customer_id")['order_id'].nunique()
Frequency=Frequency.reset_index()
```

```python
Frequency.rename(columns={"order_id":"Frequency"},inplace=True)
Frequency
```

```python
orderitems["Sales"] = orderitems["quantity"] * orderitems["list_price"]
```

```python
ordersales=orderitems.merge(orders[['order_id','customer_id']],on='order_id',how='left')
```

```python
Monetary=ordersales.groupby('customer_id')['Sales'].sum()
Monetary=Monetary.reset_index()
Monetary
```

```python
Monetary.rename(columns={'Sales':"Monetary"},inplace=True)
Monetary
```

```python
RFM=lastorders[['customer_id','Recency']].merge(Frequency,on='customer_id').merge(Monetary,on='customer_id')
RFM
```

```python
from sqlalchemy import create_engine
engine = create_engine("mysql+pymysql://root:12345@localhost:3306/capstone")
```

```python
RFM.to_sql('customer_segments',con=engine,if_exists='replace',index=False)
```
